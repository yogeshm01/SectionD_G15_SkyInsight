# Statistical Analysis

## Data Loading and Initial Setup

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/cleaned_airline_reviews.csv')
```

## Correlation Analysis

```python
cols = ['overall_rating','seat_comfort','cabin_staff_service','food_beverages','ground_service','inflight_entertainment','value_for_money']

# Convert specified columns to numeric, coercing errors to NaN
for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

corr = df[cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```python
print(df.columns)
```

## Impact of Recommendation Status

```python
grouped = df.groupby('recommended')[cols[1:]].mean()
grouped.T.plot(kind='bar')
plt.show()
```

## Overall Rating by Seat Type

```python
df.groupby('seat_type')['overall_rating'].mean().plot(kind='bar'); plt.show()
```

## Overall Rating by Type of Traveller

```python
df.groupby('type_of_traveller')['overall_rating'].mean().plot(kind='bar'); plt.show()
```

## Overall Rating by Verified Status

```python
df.groupby('verified')['overall_rating'].mean().plot(kind='bar'); plt.show()
```

### Airline Review Analysis

This notebook performs an exploratory data analysis (EDA) on airline review data to understand customer sentiment and airline performance.

### Setup and Data Loading

First, we import the necessary libraries and load the cleaned airline reviews dataset into a pandas DataFrame.

# EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../data/processed/cleaned_airline_reviews.csv')
```

### Distribution of Overall Ratings

Let's visualize the distribution of `overall_rating` to understand the general sentiment of the reviews.

```python
sns.histplot(df['overall_rating'], bins=10, kde=True)
plt.show()
```

### Calculate Average Rating per Airline

To identify top and bottom performing airlines, we calculate the average `overall_rating` for each airline.

```python
df['overall_rating'] = pd.to_numeric(df['overall_rating'], errors='coerce')
airline_avg = df.groupby('airline_name')['overall_rating'].mean().sort_values()
top10 = airline_avg.tail(10)
bottom10 = airline_avg.head(10)
```

### Top 10 Airlines by Average Rating

Let's visualize the top 10 airlines based on their average `overall_rating`.

```python
top10.plot(kind='bar'); plt.show()
```

### Bottom 10 Airlines by Average Rating

Now, let's visualize the bottom 10 airlines based on their average `overall_rating`.

```python
bottom10.plot(kind='bar'); plt.show()
```

# 🧹 Data Cleaning

```
import pandas as pd
df = pd.read_csv('../data/raw/airline_reviews.csv')
```

```
df = df.drop(columns=['Unnamed: 0'])
```

```
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')
```

```
df = df.drop(columns=['wifi_&_connectivity', 'aircraft'])

cat_cols = ['type_of_traveller', 'seat_type', 'route', 'date_flown']
for col in cat_cols:
    df[col].fillna('Unknown', inplace=True)

rating_cols = [
    'seat_comfort', 'cabin_staff_service', 'food_&_beverages',
    'ground_service', 'inflight_entertainment', 'value_for_money'
]

for col in rating_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```
df['overall_rating'] = pd.to_numeric(df['overall_rating'], errors='coerce')
df['review_date'] = pd.to_datetime(df['review_date'], errors='coerce')
df['date_flown'] = pd.to_datetime(df['date_flown'], errors='coerce')
```

```
df['recommended'] = df['recommended'].str.lower().str.strip()
df['is_recommended'] = df['recommended'].map({'yes':1, 'no':0})
```

```
df = df.dropna(subset=['overall_rating'])
```

```
df.to_csv('../data/processed/cleaned_airline_reviews.csv', index=False)
```
